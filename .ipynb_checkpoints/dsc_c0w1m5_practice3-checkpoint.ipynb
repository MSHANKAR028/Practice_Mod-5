# Practice for Aggregating, Reshaping, and Combining Dataframes

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

1. Load customer demographics and claims information into separate dataframes:
- customer demographics should be loaded into `cust_df` from 'cust_demographics.csv'
- claims should be loaded into `claims_df` from 'claims.csv'

```python
cust_df = pd.read_csv("cust_demographics.csv")
claims_df = pd.read_csv("claims.csv")
```

2. Inspect the DataFrames to get an understanding of the data, dtypes, and column structure

```python
cust_df.info()
print("\n" + "="*40 + "\n")

# Inspect claims data
claims_df.info()
```

3. Check for duplicates and then remove them:

```python
# Remove duplicates from both dataframes
cust_df.drop_duplicates(inplace=True)
claims_df.drop_duplicates(inplace=True)
```

4. Join the dataframes and save to a dataframe `combined_df`. Identify what column(s) you will join on and then use an inner join. Drop any redundant columns resulting from merge:

```python
print("Customer columns:", cust_df.columns)
print("Claims columns:", claims_df.columns)
```

```python
# Join using the two different column names
combined_df = pd.merge(cust_df, claims_df, left_on='CUST_ID', right_on='customer_id', how='inner')

# Drop the redundant 'customer_id' column so we don't have duplicates
combined_df.drop(columns=['customer_id'], inplace=True)

# Take a look at the beautiful combined result
combined_df.head()
```

5. Identify cleaning tasks in the `claim_amount` column and execute them:

```python
# Strip the dollar sign and convert to float numbers
combined_df['claim_amount'] = combined_df['claim_amount'].astype(str).str.replace('$', '', regex=False).astype(float)

# Check the results
combined_df['claim_amount'].head()
```

6. Clean fradulent and convert to appropriate numeric values:

```python
# Map 'Yes'/'No' strings to 1 and 0
fraud_map = {'Yes': 1, 'No': 0}
combined_df['fraudulent'] = combined_df['fraudulent'].map(fraud_map)

# Check the values to make sure it worked
combined_df['fraudulent'].value_counts()
```

7. Use an appropriate plot to visualize the the mean claim amount for each incidence cause:

```python
# Group by cause and calculate the mean amount
cause_means = combined_df.groupby('incident_cause')['claim_amount'].mean()

# Plot the bar chart
cause_means.plot(kind='bar', color='teal', figsize=(10, 5))
plt.title('Mean Claim Amount by Incident Cause')
plt.xlabel('Incident Cause')
plt.ylabel('Average Claim Amount ($)')
plt.xticks(rotation=45)
plt.show()
```

8. Plot the distribution of claim amounts: use a histogram.

```python
# Plot a histogram to see the distribution shape
combined_df['claim_amount'].hist(bins=30, color='skyblue', edgecolor='black', figsize=(10, 5))
plt.title('Distribution of Claim Amounts')
plt.xlabel('Claim Amount ($)')
plt.ylabel('Frequency')
plt.show()
```

9. There is a bimodal distribution here. Create a binary column called "claim_size" that is "large" when the claim amount is greater than or equal to 10000 and "small" otherwise. We will use this binary categorical later.

```python
# Create the binary categorical flag using np.where
combined_df['claim_size'] = np.where(combined_df['claim_amount'] >= 10000, 'large', 'small')

# Check our new categories
combined_df['claim_size'].value_counts()
```

10. Investigate the fraudulent claim rate by state:
- get the rate of fraudulent claims by state sorted in descending order. Save this to a Series `sorted_rates`
- get the states with the top 15 fraudulent claim rates and save this list of state to `state_names`.

```python
# 1. Get the rate of fraud by state, sorted from highest to lowest
sorted_rates = combined_df.groupby('State')['fraudulent'].mean().sort_values(ascending=False)

# 2. Extract just the names of the top 15 states
state_names = sorted_rates.head(15).index.tolist()

# Let's see your lists!
print("Top 5 Sorted Rates:\n", sorted_rates.head())
print("\nTop 15 State Names:\n", state_names)
```

11. It would be interesting to see whether there is a differentiation of fraud rates by claim size in each state. First compute this by using a pivot table. Display results from the states with the top 15 highest fraudulent claim rates.


```python
# Create the pivot table
pivot_fraud = combined_df.pivot_table(values='fraudulent', index='State', columns='claim_size', aggfunc='mean')

# Filter for the top 15 states we found in step 10
top_15_pivot = pivot_fraud.loc[state_names]
top_15_pivot
```

12. Conduct a groupby/aggregation(s) by State and claim size. This time you will compute the fraudulent claim rate as well as the number of examples in each category and subcategory. The result should be presented in a single DataFrame.

*Hint*: you can compute each aggregation separately and then concatenate **or** you can look up how to perform multiple aggregations with the .agg() method.

```python
# Group by State and claim_size, then calculate mean and count
state_claim_groups = combined_df.groupby(['State', 'claim_size'])['fraudulent'].agg(['mean', 'count'])

# Filter the results so it only shows the top 15 states we found
top_15_groups = state_claim_groups.loc[state_names]
top_15_groups
```

There are not too many examples within each subgroup -- but it is plausible that there are enough here to differentiate between fraud behavior for small and large claims in many of the states.


13. Plot fraudulent rates amongst small and large claim sizes in the 15 states with the top average fraudulent rates:
- put this in a single bar chart with large and small claim sizes overlayed on top of each other
- use blue for large claim and orange for small claim, set transparency parameter to 0.4.
- make sure that you label plots appropriately and display legends.

*Hint*: selecting the appropriate column and unstacking the results of your groubpy on State and claim size might be helpful here

```python
# Plot large claims in blue
ax = top_15_pivot['large'].plot(kind='bar', color='blue', alpha=0.4, label='Large Claims', figsize=(12, 6))

# Overlay small claims in orange on the same 'ax'
top_15_pivot['small'].plot(kind='bar', color='orange', alpha=0.4, label='Small Claims', ax=ax)

# Add titles, labels, and the legend
plt.title('Fraudulent Rates by Claim Size in Top 15 States')
plt.xlabel('State')
plt.ylabel('Fraud Rate')
plt.xticks(rotation=45)
plt.legend()
plt.show()
```

14. Just because its a lot of fun and good for you, melt the pivot table with average fraudulent rates (pivoted on State and claim size) for the states with top 15 highest overall fradulent rates. We have recalculated the table that is to be melted below from `combined_df` and have stored it in `widestateclaims`:
- make sure that the index is still State (look up what option for melt you need)
- the value should be named 'fraud_rate'
- save to `melted_claims`

```python
# 1. Recreate the wide-format pivot table first for the top 15 states,
widestateclaims = combined_df.pivot_table(values='fraudulent', index='State', columns='claim_size', aggfunc='mean').loc[state_names]

# 2. Reset the index so 'State' becomes a column we can melt on, then melt it: 
melted_claims = widestateclaims.reset_index().melt(id_vars='State', value_name='fraud_rate')

# 3. Set the index back to 'State' as requested
melted_claims.set_index('State', inplace=True)

# Preview the melted long-format data
melted_claims.head()
```
